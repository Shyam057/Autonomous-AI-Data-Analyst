```python
import os
from dotenv import load_dotenv
from typing import Annotated, TypedDict

from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langgraph.prebuilt import ToolNode, tools_condition

load_dotenv("/home/susan/Shyam/Autonomous-AI-Data-Aanalyst/.env")

api_key = os.getenv("GROQ_API_KEY")
if not api_key:
    raise RuntimeError("GROQ_API_KEY was not loaded.")

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    groq_api_key=api_key,
    temperature=0,
)
```

```python
from typing import Annotated, TypedDict
from langgraph.graph.message import add_messages
from langgraph.prebuilt import ToolNode, tools_condition

class AnalystState(TypedDict):
    messages: Annotated[list, add_messages]


tools = [
    dataset_overview,
    column_statistics,
    churn_analysis,
    visualize_column,
]

llm_with_tools = llm.bind_tools(tools)
```

```python
def call_llm(state: AnalystState):
    response = llm_with_tools.invoke(state["messages"])

    return {
        "messages": [response]
    }
```

```python
graph_builder = StateGraph(AnalystState)

graph_builder.add_node("llm", call_llm)

graph_builder.add_edge(START, "llm")
graph_builder.add_edge("llm", END)

graph = graph_builder.compile()
```

```python
result = graph.invoke({
    "messages": [
        HumanMessage(
            content="How many customers are in the dataset?"
        )
    ]
})

result
```
